```python
%%sql

CREATE OR REPLACE TABLE dbo.t1 USING DELTA AS (
    SELECT * FROM t2 FULL OUTER JOIN t3 LIMIT 10000
)
```

```python
%%sql

optimize dbo.t1 vorder
```

```python
%%sql

select age, countryOrRegion, count(1) as total from dbo.t1 group by age, countryOrRegion sort by 3 desc
```

```python
%%sql

-- Additive schema change: create analytics table t5
CREATE TABLE IF NOT EXISTS dbo.t5 USING DELTA AS (
    SELECT 
        t1.AGE,
        t1.SEX,
        t1.BMI,
        t1.countryOrRegion,
        t3.holidayName,
        CURRENT_TIMESTAMP() as created_timestamp
    FROM dbo.t1 t1
    INNER JOIN dbo.t3 t3 ON t1.countryOrRegion = t3.countryOrRegion
)
```

```python
%%sql

SELECT COUNT(*) as t5_row_count FROM dbo.t5
```
